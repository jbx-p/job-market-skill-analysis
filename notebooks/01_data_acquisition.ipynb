### Load and inspect the main postings file

```python
import pandas as pd

postings = pd.read_csv("../data/raw/postings.csv")
print(postings.shape)
print(postings.columns.tolist())
postings.head()
```

## Phase 2 — Filter to Relevant Roles and Markets

### Step 1 — filter to marketing/data analyst titles

```python
title_keywords = ["marketing analyst", "data analyst", "marketing data analyst",
                    "digital analyst", "insights analyst", "business analyst"]

pattern = "|".join(title_keywords)
relevant = postings[postings["title"].str.contains(pattern, case=False, na=False)]

print(f"Relevant postings: {len(relevant):,} of {len(postings):,} total")
print(relevant["title"].value_counts().head(20))
```

### Step 2 — filter to your target countries

```python
target_locations = ["Canada", "Australia"]
location_pattern = "|".join(target_locations)

relevant_geo = relevant[relevant["location"].str.contains(location_pattern, case=False, na=False)]
print(f"Relevant postings in target countries: {len(relevant_geo):,}")
print(relevant_geo["location"].value_counts().head(20))
```

```python
if len(relevant_geo) < 500:
    print("Canada/Australia sample too small — broadening to include US for statistical power")
    target_locations = ["Canada", "Australia", "United States"]
    location_pattern = "|".join(target_locations)
    relevant_geo = relevant[relevant["location"].str.contains(location_pattern, case=False, na=False)]
    print(f"Broadened sample: {len(relevant_geo):,}")
```
